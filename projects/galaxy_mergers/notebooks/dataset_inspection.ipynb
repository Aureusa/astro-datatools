```python
DATASET_FILEPATH = "/home/s4861264/project_data/merger_project"
TXT_BASE_NAME = "L200m6_1e8"

# Read .txt
def read_txt(snapshot):
    filepath = f"{DATASET_FILEPATH}/{TXT_BASE_NAME}_{snapshot}.txt"
    lines = []
    with open(filepath, "r") as f:
        data = f.read()
        # '1.559654000000000000e+06' per line to int
        lines = [int(float(line)) for line in data.splitlines()]
    return lines

def get_all_experiments():
    import os
    all_exps = []
    for filename in os.listdir(DATASET_FILEPATH):
        if filename.startswith(TXT_BASE_NAME) and filename.endswith(".txt"):
            snapshot = filename[len(TXT_BASE_NAME)+1:-4]
            all_exps.append(snapshot)

    # Sort the experiments before returning
    all_exps.sort()
    return all_exps
```

```python
all_experiments = get_all_experiments()
print(all_experiments)
print(f"Total number of experiments: {len(all_experiments)}")
```

```python
total_num_galaxies = 0
unique_galaxies = {}

for exp in all_experiments:
    data = read_txt(exp)
    num_galaxies = len(data)
    for galaxy in data:
        if galaxy not in unique_galaxies:
            unique_galaxies[galaxy] = {"count": 0, "snapshots": []}
        unique_galaxies[galaxy]["count"] += 1
        unique_galaxies[galaxy]["snapshots"].append(exp)
    total_num_galaxies += num_galaxies
```

```python
print(f"Total number of galaxies: {total_num_galaxies}")
print(f"Unique galaxies: {len(unique_galaxies)}")
count = 0
for galaxy, info in unique_galaxies.items():
    if info['count'] < 4:
        continue
    print(f"Galaxy ID: {galaxy}, Count: {info['count']}, Snapshots: {info['snapshots']}")
    count += 1
    if count >= 10:
        break

max_count = 0
for galaxy, info in unique_galaxies.items():
    if info['count'] > max_count:
        max_count = info['count']
    if info['count'] == max_count:
        print(f"Galaxy with maximum snapshots: {galaxy}, Count: {info['count']}, Snapshots: {info['snapshots']}")
print(f"Maximum count of snapshots for a single galaxy: {max_count}")
```
